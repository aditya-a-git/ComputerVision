```python
from google.colab import files
from pathlib import Path
import shutil, yaml
upload = files.upload()
archive = next(iter(upload))
root = Path('data/colab').resolve()
shutil.unpack_archive(archive, root)
config = root / 'data.yaml'
spec = yaml.safe_load(config.read_text())
spec['path'] = str(root)
config.write_text(yaml.safe_dump(spec))
```

```python
import torch, subprocess, sys
subprocess.run([sys.executable, '-m', 'scripts.train', '--data', str(config), '--device', '0' if torch.cuda.is_available() else 'cpu'], check=True)
```

```python
checkpoint = max(Path('runs/train').glob('*/weights/best.pt'),key=lambda p:p.stat().st_mtime)
shutil.copy(checkpoint,'rdd_best.pt')
files.download('rdd_best.pt')
```
